```python
import os
import random
import shutil

# Пути к исходным данным
input_images = "/home/zsv/Projects/itmo/dataset/images"  # Исходные изображения
input_labels = "/home/zsv/Projects/itmo/dataset/labels"  # Исходные аннотации

# Пути для выходных данных
output_train_images = "/home/zsv/Projects/itmo/dataset/train/images"
output_train_labels = "/home/zsv/Projects/itmo/dataset/train/labels"
output_val_images = "/home/zsv/Projects/itmo/dataset/val/images"
output_val_labels = "/home/zsv/Projects/itmo/dataset/val/labels"

# Создание необходимых директорий
os.makedirs(output_train_images, exist_ok=True)
os.makedirs(output_train_labels, exist_ok=True)
os.makedirs(output_val_images, exist_ok=True)
os.makedirs(output_val_labels, exist_ok=True)

# Список всех изображений
image_files = [f for f in os.listdir(input_images) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# Перемешивание списка файлов
random.seed(1337)  # Для воспроизводимости
random.shuffle(image_files)

# Разделение на train и val (например, 90% train, 10% val)
train_ratio = 0.9
train_size = int(len(image_files) * train_ratio)
train_files = image_files[:train_size]
val_files = image_files[train_size:]

# Функция для копирования файлов
def copy_files(file_list, src_images, src_labels, dest_images, dest_labels):
    for file_name in file_list:
        # Копирование изображений
        src_image_path = os.path.join(src_images, file_name)
        shutil.copy(src_image_path, dest_images)
        
        # Копирование соответствующих аннотаций
        label_name = os.path.splitext(file_name)[0] + ".txt"
        src_label_path = os.path.join(src_labels, label_name)
        if os.path.exists(src_label_path):
            shutil.copy(src_label_path, dest_labels)

# Копирование train данных
copy_files(train_files, input_images, input_labels, output_train_images, output_train_labels)

# Копирование val данных
copy_files(val_files, input_images, input_labels, output_val_images, output_val_labels)

print("Данные успешно разделены!")
```

```python
import os
import shutil

# Пути к данным
input_labels_train = "/home/zsv/Projects/itmo/dataset/labels/train"  # Исходные аннотации
output_train_labels = "/home/zsv/Projects/itmo/dataset/train/labels"  # Аннотации для train
output_val_labels = "/home/zsv/Projects/itmo/dataset/val/labels"  # Аннотации для val

# Создание целевых директорий
os.makedirs(output_train_labels, exist_ok=True)
os.makedirs(output_val_labels, exist_ok=True)

# Функция для копирования аннотаций
def copy_labels(image_folder, label_folder, output_label_folder):
    for image_file in os.listdir(image_folder):
        if image_file.lower().endswith(('.png', '.jpg', '.jpeg')):
            label_name = os.path.splitext(image_file)[0] + ".txt"
            src_label_path = os.path.join(label_folder, label_name)
            if os.path.exists(src_label_path):  # Проверяем, существует ли аннотация
                shutil.copy(src_label_path, output_label_folder)
            else:
                print(f"Warning: No label found for image {image_file}")

# Копирование аннотаций для train
train_images_folder = "/home/zsv/Projects/itmo/dataset/train/images"
copy_labels(train_images_folder, input_labels_train, output_train_labels)

# Копирование аннотаций для val
val_images_folder = "/home/zsv/Projects/itmo/dataset/val/images"
copy_labels(val_images_folder, input_labels_train, output_val_labels)

print("Аннотации успешно разделены!")
```

```python
import cv2
from ultralytics import YOLO

# Путь к модели
checkpoint_path = '/home/zsv/Projects/itmo/model/helmet_run_100_epoches/weights/best.pt'

# Загрузка модели
model = YOLO(checkpoint_path)

# Захват видео с веб-камеры
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Ошибка: не удалось открыть веб-камеру")
    exit()

# Параметры для записи видео
output_path = '/home/zsv/Projects/itmo/itmo_rezult_mishs.avi'
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(cv2.CAP_PROP_FPS) or 30)  # Если FPS не определён, ставим 30

# Инициализация видеозаписи
fourcc = cv2.VideoWriter_fourcc(*'XVID')  # Кодек XVID
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

while True:
    ret, frame = cap.read()
    if not ret:
        print("Ошибка: не удалось считать кадр")
        break

    # Выполнение предсказания
    results = model.predict(frame)

    # Отображение результатов
    for result in results:
        for box in result.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            conf = box.conf[0].item()
            cls = int(box.cls[0].item())
            label = f"{model.names[cls]} {conf:.2f}"

            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Запись кадра в файл
    out.write(frame)

    cv2.imshow("YOLO Inference", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

# Освобождение ресурсов
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Видео сохранено в {output_path}")
```
